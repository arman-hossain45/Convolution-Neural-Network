# Question 03: [Marks 5]

## Discuss the vanishing gradient problems in deep neural networks. Explain why they occur during backpropagation, how activation functions contribute to this issue, and suggest at least three techniques to overcome these problems.


## Write Answer 03:
The vanishing gradient problem occurs in deep neural networks when the gradients  become extremely small as they are propagated backward through many layers. During backpropagation, gradients are calculated using the chain rule, which involves multiplying many derivatives together. If these derivatives are small  their repeated multiplication causes the gradient to shrink exponentially as it moves toward earlier layers. As a result, the weights in the initial layers receive almost no updates, making learning very slow or even stopping it completely.


Activation functions like sigmoid and relu make values very small by squeezing them into a limited range. Because of this, their derivatives are also small  During backpropagation  these small numbers get multiplied again and again across many layers, so the gradient becomes almost zero. As a result, earlier layers don’t learn properly, and the network struggles to improve.

There are maninly 3 technoiq to improve this

1.Reduce complexity

2.batchNormalization

3.ReLU

## Describe the architecture of an Artificial Neural Network (ANN) in detail. Explain the role of each component (neurons, weights, bias, activation functions). Also discuss how increasing the number of hidden layers and neurons affects model performance and overfitting.


## Write Answer 04:
An Artificial Neural Network  is a model made of layers of connected neurons that process data step by step. It has three main parts  an input layer takes features as a input, one or more hidden layers  to learn patterns, and an output layer which gives the final prediction. Each neuron receives inputs, multiplies them by weights , adds a bias , and then passes the result through an activation function like ReLU or sigmoid to introduce non-linearity. Mathematically, a neuron works like:

z=W⋅x+b

a=f(z)

where W = weights, b = bias, and f = activation function.

Increasing the number of hidden layers and neurons makes the model more powerful, allowing it to learn complex patterns and improve accuracy. However, if the network becomes too large, it may overfit, meaning it memorizes the training data but performs poorly on new data. So, a balance is important: deeper and wider networks improve learning, but too much complexity can reduce generalization.

comparison

MaxPooling focuses on the strongest feature in a region by selecting the maximum value. This means if an important pattern like an edge or a bright spot exists anywhere in that small window, MaxPooling will keep it. Because of this, it is very effective for feature detection, especially in image tasks where the presence of a feature matters more than its exact position.

AveragePooling, on the other hand, takes the average of all values in the region. Instead of focusing on the strongest feature, it tries to keep a general summary of that area. This makes the output smoother, but it may lose sharp or important details. So, while it preserves overall information, it is less effective than MaxPooling for detecting strong patterns.

Pooling is important because of

1. Reduces Overfitting
Pooling reduces the size of the feature maps, which means the model has fewer parameters to learn. With fewer details, the model cannot memorize noise or unnecessary patterns from the training data. Instead, it learns more general and important features, which helps it perform better on new, unseen data.

2. Reduces Computational Complexity
When pooling reduces the feature map size (for example, from 28×28 to 14×14), the number of computations in the next layers becomes much smaller. This leads to:

Faster training

Less memory usage

More efficient models

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
from torch.utils.data import Dataset,DataLoader
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(42)
```

```python
# check the gpu
device= torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device :{device}")
```

```python
df=pd.read_csv("fmnist_small.csv")
df.head(5)
```

```python


# Create a 4x4 grid of images
fig, axes = plt.subplots(4, 4, figsize=(10, 10))
fig.suptitle("First 16 Images", fontsize=16)

# Plot the first 16 images from the dataset
for i, ax in enumerate(axes.flat):
    img = df.iloc[i, 1:].values.reshape(28, 28)  # Reshape to 28x28
    ax.imshow(img)  # Display in grayscale
    ax.axis('off')  # Remove axis for a cleaner look
    ax.set_title(f"Label: {df.iloc[i, 0]}")  # Show the label

plt.tight_layout(rect=[0, 0, 1, 0.96])  # Adjust layout to fit the title
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop('label',axis=1)
y=df['label']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train=x_train/255.0
x_test=x_test/255.0
```

```python
class CustomDataset(Dataset):
  def __init__(self,features,labels):
    # convert to pytorch to tensor
    self.features=torch.tensor(features.values,dtype=torch.float32).reshape(-1,1,28,28)
    self.labels=torch.tensor(labels.values,dtype=torch.long)
  def __len__(self):
    return len(self.features)

  def __getitem__(self, index):
     return self.features[index],self.labels[index]
```

```python
train_dataset=CustomDataset(x_train,y_train)
test_dataset =CustomDataset(x_test,y_test)
```

```python
train_loader= DataLoader(train_dataset,batch_size=32,shuffle=True,pin_memory=True)
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=True,pin_memory=True)
```

Trainng loop in details

```python

class MyNN(nn.Module):
    def __init__(self, input_features):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(input_features, 32, kernel_size=3, padding='same'),
            nn.ReLU(),
            nn.BatchNorm2d(32),
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(32, 64, kernel_size=3, padding='same'),
            nn.ReLU(),
            nn.BatchNorm2d(64),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64*7*7, 128),
            nn.ReLU(),
            nn.Dropout(p=0.4),

            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(p=0.4),

            nn.Linear(64, 10)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)

        return x


learning_rate = 0.01
epochs = 100
model = MyNN(1)

model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=learning_rate, weight_decay=1e-4)

```

```python
for epoch in range(epochs):

  total_epoch_loss = 0

  for batch_features, batch_labels in train_loader:

    # move data to gpu
    batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)

    # forward pass
    outputs = model(batch_features)

    # calculate loss
    loss = criterion(outputs, batch_labels)

    # back pass
    optimizer.zero_grad()
    loss.backward()

    # update grads
    optimizer.step()

    total_epoch_loss = total_epoch_loss + loss.item()

  avg_loss = total_epoch_loss/len(train_loader)
  print(f'Epoch: {epoch + 1} , Loss: {avg_loss}')
```

```python
model.eval()
```

```python
total = 0
correct = 0

with torch.no_grad():

  for batch_features, batch_labels in test_loader:

    # move data to gpu
    batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)

    outputs = model(batch_features)

    _, predicted = torch.max(outputs, 1)

    total = total + batch_labels.shape[0]

    correct = correct + (predicted == batch_labels).sum().item()

print(correct/total)
```
